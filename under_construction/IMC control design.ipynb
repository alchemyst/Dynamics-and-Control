```python
import sympy
```

## Closed loop transfer functions

```python
y_sp, e, p, yprime, d, y, ytilde, yminy, Gcstar, G, Gtilde = sympy.symbols('y_sp, e, p, yprime, d, y, ytilde, yminy, Gcstar, G, Gtilde')
```

```python
blocks = [p - Gcstar*e, yprime - G*p, ytilde - Gtilde*p]
sums = [e - (ysp - yminy), y - (d + ytilde), yminy - (y - ytilde)]
unknowns = [e, p, yprime, y, ytilde, yminy]
```
