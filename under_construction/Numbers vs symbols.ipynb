## Python functions aren't equations

Students who encounter Python for the first time in the context of solving mostly mathematical problems often allow themselves to think of Python functions as being quite similar to mathematical objects:

```python
def f(x):
    return x**2 + 1
```

```python
f(2)
```

However, Python functions can do much more than "just" mathematics:

```python
def say_hello(name):
    print("Hello", name)
```

```python
say_hello("Carl")
```

Clearly, this function has nothing to do with mathematics!

### Using sympy variables in Python functions

sympy allows us to define symbolic variables and pass them through to Python functions, which if they are doing only "normal" math, works quite well:

```python
import sympy
sympy.init_printing()
```

```python
x = sympy.Symbol('x')
```

```python
f(x)
```
