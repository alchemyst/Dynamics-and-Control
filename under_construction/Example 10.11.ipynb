```python
s = sympy.symbols('s')
K_c = sympy.symbols('K_c', real=True)
```

```python
Gc = K_c
Gv = 2
Gm = 0.25
Gp = 4*sympy.exp(-s)/(5*s + 1)
```

```python
Gol = Gc*Gv*Gm*Gp
Gcl = Gol/(1 + Gol)
```

```python
CE = sympy.denom(Gcl.cancel())
```

```python
cepoly = sympy.poly(
    sympy.nsimplify(
        sympy.numer(
            CE.subs(
                sympy.exp(-s), (1 - 0.5*s)/(1 + 0.5*s)
            ).cancel()
        )), s)
```

```python
r = routh(cepoly)
```

```python
sympy.solve([item < 0 for item in r[:, 0]], K_c)
```
