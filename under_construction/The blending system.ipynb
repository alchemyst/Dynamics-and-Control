# The blending system
<img src="../../assets/blending_system.png" width="50%">

Model:

\begin{align}
\frac{\mathrm{d}(V\rho)}{\mathrm{d}t} &= w_1 + w_2 - w \\
\frac{\mathrm{d}(V\rho x)}{\mathrm{d}t} &= w_1x_1 + w_2x_2 - wx \\
\end{align}

```python
V = 2  # m3   - "constant liquid holdup"
ρ = 900  # kg/m3
```

```python
w1bar = 500  # kg/min
w2bar = 200  # kg/min
```

```python
x1 = 0.4
x2 = 0.75
```

Solve for steady state

```python
import sympy
sympy.init_printing()
```

```python
w, w1, w2, x, t = sympy.symbols('w, w1, w2, x, t')
```

```python
sol = sympy.solve([w1 + w2 - w, w1*x1 + w2*x2 - w*x], [w, x])
```

```python
[(wbar, xbar)] = sol
```

```python
wbar, xbar
```

Solve the differential equation - note we need to redefine x here

```python
x = sympy.Function('x')
```

```python
diffeq = sympy.diff(x(t), t) - (w1*x1 + w2*x2 - w*x(t)).subs({w: wbar})
```

```python
general_solution = sympy.dsolve(diffeq)
```
