```python
import sympy
sympy.init_printing()
```

```python
%matplotlib inline
```

```python
t, s = sympy.symbols('t, s')
```

```python
u = 1/s**2*(1 - 2*sympy.exp(-s) + sympy.exp(-2*s))
```

First order

```python
G = 1/(s + 1)
```

```python
y = G*u
```

```python
def plotresponse(G):
    sympy.plot(sympy.inverse_laplace_transform(G*u, s, t))
plotresponse(G)
```

Second order
=======

```python
tau, zeta = sympy.symbols('tau, zeta', real=True, positive=True)
G2 = 1/(tau**2*s**2 + 2*tau*zeta*s + 1)
```

Overdamped
---------

```python
response = G2.subs({zeta: 2, tau: 1})*u
```

```python
response
```

```python
target = response.apart()
```

```python
inverse = sum(sympy.inverse_laplace_transform(term, s, t) for term in target.as_ordered_terms())
```
