```python
def fin(t):
    while time.time() - tstart < t:
        time.sleep(0.5)
    if t < 5:
        return 0
    else:
        return 1
```

```python
for t in range(0, 10):
    print(t, tstart - time.time(), fin(t))
```

```python
def f(t, x):
#     print(f'Called with t={t} and x={x}')
    fout = 2*x
    return fin(t) - fout
```

```python
def my_solve_ivp(f, ts, x0):
    dt = ts[1]
    x = x0
    result = []
    for t in ts:
        dxdt = f(t, x)
        x += dxdt*dt
        result.append(x)
    return result    
```

```python
import numpy
```

```python
ts = numpy.linspace(0, 10, 100)
```

```python
xs = my_solve_ivp(f, ts, 1)
```
