```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
import numpy as np
```

```python
fig, [axgain, axphase] = plt.subplots(2, 1)

for dt in [0.001, 0.01, 0.1, 1, 2]:
    t0 = 100
    ts = np.arange(0, t0, dt)
    impulse = np.exp(-ts)
    f = 1/dt
    fft = np.fft.rfft(impulse)*dt
    oms = np.fft.rfftfreq(len(impulse), dt)*2*np.pi
    Ar2 = np.abs(fft)
    phi = np.unwrap(np.angle(fft))
    axgain.loglog(oms, Ar2)
    axgain.text(oms[-1], Ar2[-1], dt)
    axphase.semilogx(oms, phi)
    axphase.text(oms[-1], phi[-1], dt)


oms = np.logspace(-2, 2)
s = 1j*oms
G = 1/(s + 1)
axgain.loglog(oms, np.abs(G), linewidth=5, alpha=0.5)
axphase.semilogx(oms, np.unwrap(np.angle(G)), linewidth=5, alpha=0.5)
```
